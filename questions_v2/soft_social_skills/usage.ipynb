## 4. Trait Scores Visualization

```python
# Bar chart of trait scores
scores = report['trait_scores']
traits = list(scores.keys())
values = [scores[t]['score'] for t in traits]
labels = [scores[t]['label_en'] for t in traits]

# Simple text bar chart
print("\nTrait Scores (0-100)")
print("-" * 60)
for label, val in sorted(zip(labels, values), key=lambda x: -x[1]):
    bar = '█' * int(val / 2.5)
    print(f"  {label:<28} {val:5.1f}  {bar}")
```

## 6. Batch Evaluation (Multiple Students)

```python
# Simulate 5 students and batch evaluate
batch_request = {
    "students": [],
    "generate_group_recommendations": True
}

names = ['Sara Ahmed', 'Omar Hassan',
         'Layla Mahmoud', 'Youssef Ali', 'Nour Ibrahim']
for name in names:
    ag = random.choice(['6-9', '10-14', '15-18'])
    ages = {'6-9': [6, 7, 8, 9], '10-14': [10, 11,
                                           12, 13, 14], '15-18': [15, 16, 17, 18]}
    qs = QUESTION_BANK[ag]
    selected = random.sample(qs, min(20, len(qs)))
    answers = {str(q['id']): random.randint(0, 3) for q in selected}

    batch_request['students'].append({
        'student_info': {'name': name, 'student_id': f'NGEN-{random.randint(1000,9999)}', 'age': random.choice(ages[ag])},
        'age_group': ag,
        'answers': answers
    })

result = evaluate_batch_api(batch_request)
print(
    f"Batch evaluation: {result['total_evaluated']} students evaluated, {result['total_errors']} errors")
print(
    f"\nGroup recommendations generated: {result['group_recommendations'] is not None}")
```

```python
result['reports'][0].keys()
```

```python
# Print summary for each student
for r in result['reports']:
    top = max(r['trait_scores'].items(), key=lambda x: x[1]['score'])
    ws = r['behavioral_dimensions']['work_style']['primary_label_en']
    pace = r['learning_profile']['learning_pace']['primary']
    group = r['learning_profile']['group_placement']['primary_label_en']
    # print(group)
    print(
        f"\n  {r['meta']['student_name']} (age {r['meta']['age']}, group {r['meta']['age_group']})")
    print(
        f"    Top: {top[1]['label_en']} ({top[1]['score']:.0f}), Style: {ws}, Pace: {pace}, Group: {group} ")
```

## 7. API Function Test

```python
# Test the API with a manual request
api_request = {
    "student_info": {
        "name": "Test API Student",
        "student_id": "API-001",
        "age": 15
    },
    "age_group": "15-18",
    "answers": {str(i): random.randint(0, 3) for i in range(1, 26)}
}

api_result = evaluate_student_api(api_request)
print(f"API Success: {api_result['success']}")
if api_result['success']:
    r = api_result['report']
    print(f"Student: {r['meta']['student_name']}")
    print(f"Age Group: {r['meta']['age_group']}")
    print(f"Questions: {r['meta']['total_questions_answered']}")
    print(f"\nTop Strengths:")
    for s in r['strengths'][:3]:
        print(f"  - {s['label_en']}: {s['score']:.1f}")
else:
    print(f"Error: {api_result['error']}")
```

```python
# Test error handling
print("\n--- Error Handling Tests ---")

# Missing field
r1 = evaluate_student_api(
    {"student_info": {"name": "X", "student_id": "1", "age": 8}})
print(f"Missing age_group: {r1['error']}")

# Invalid age group
r2 = evaluate_student_api({"student_info": {
                          "name": "X", "student_id": "1", "age": 8}, "age_group": "99-100", "answers": {}})
print(f"Invalid age_group: {r2['error']}")

# Too few answers
r3 = evaluate_student_api({"student_info": {
                          "name": "X", "student_id": "1", "age": 8}, "age_group": "6-9", "answers": {"1": 0}})
print(f"Too few answers: {r3['error']}")

print("\n✅ All error cases handled correctly!")
```
