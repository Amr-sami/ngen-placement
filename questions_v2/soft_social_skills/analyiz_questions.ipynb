```python
import json
from collections import Counter


def load_questions(filename="questions_en_ar.json"):
    """Load questions from JSON file."""
    try:
        with open(filename, 'r', encoding='utf-8') as file:
            data = json.load(file)
        return data
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
        return None
    except json.JSONDecodeError as e:
        print(f"Error: Invalid JSON format - {e}")
        return None


def analyze_questions(data):
    """Provide comprehensive analysis of the questions dataset."""

    if not data:
        return "No data to analyze"

    analysis = {
        "total_questions": len(data),
        "age_groups": Counter(),
        "concepts_frequency": Counter(),
        "avg_options_per_question": 0,
        "questions_with_arabic": 0,
        "all_concepts": set()
    }

    total_options = 0

    for item in data:
        # Count age groups
        age_group = item.get("age_group", "Unknown")
        analysis["age_groups"][age_group] += 1

        # Count concepts
        concepts = item.get("concepts_en", [])
        for concept in concepts:
            analysis["concepts_frequency"][concept] += 1
            analysis["all_concepts"].add(concept)

        # Count options
        options = item.get("options_en", [])
        total_options += len(options)

        # Check for Arabic translation
        if item.get("question_ar"):
            analysis["questions_with_arabic"] += 1

    analysis["avg_options_per_question"] = total_options / \
        len(data) if data else 0
    analysis["all_concepts"] = list(analysis["all_concepts"])

    return analysis


def print_analysis_report(analysis):
    """Print a formatted analysis report."""

    print("=" * 50)
    print("📊 QUESTIONS DATASET ANALYSIS REPORT")
    print("=" * 50)

    print(f"\n📌 Total Questions: {analysis['total_questions']}")
    print(f"📌 Questions with Arabic: {analysis['questions_with_arabic']}")
    print(
        f"📌 Avg Options per Question: {analysis['avg_options_per_question']:.1f}")

    print("\n📂 Age Groups Distribution:")
    for age, count in analysis['age_groups'].items():
        print(f"   • {age}: {count} questions")

    print("\n🧠 Concepts Frequency:")
    for concept, count in analysis['concepts_frequency'].most_common():
        print(f"   • {concept}: {count} occurrences")

    print("\n📝 All Unique Concepts:")
    print(f"   {', '.join(analysis['all_concepts'])}")
    print("=" * 50)


# Main execution
if __name__ == "__main__":
    # Load the data
    questions = load_questions("questions_en_ar.json")

    if questions:
        # Run analysis
        analysis = analyze_questions(questions)

        # Print report
        print_analysis_report(analysis)

        # Optional: Export analysis to JSON
        with open("analysis_report.json", 'w', encoding='utf-8') as f:
            json.dump(analysis, f, indent=4, ensure_ascii=False)
```

```python
import json
import random


def load_questions(filename="questions_en_ar.json"):
    """Load questions from JSON file."""
    try:
        with open(filename, 'r', encoding='utf-8') as file:
            data = json.load(file)
        return data
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
        return None
    except json.JSONDecodeError as e:
        print(f"Error: Invalid JSON format - {e}")
        return None


def get_random_questions_by_age(data, age_group, count=2):
    """
    Get random questions for a specific age range.

    Parameters:
    -----------
    data : list
        List of question dictionaries
    age_group : str
        Target age group (e.g., "6-9", "10-12", "13-15")
    count : int
        Number of random questions to return (default: 2)

    Returns:
    --------
    list : Random questions matching the age group
    """

    if not data:
        print("Error: No data provided.")
        return []

    # Filter questions by age group
    filtered_questions = [
        question for question in data
        if question.get("age_group") == age_group
    ]

    # Check if we found any questions
    if not filtered_questions:
        print(f"No questions found for age group: {age_group}")
        print(f"Available age groups: {get_available_age_groups(data)}")
        return []

    # Check if we have enough questions
    if len(filtered_questions) < count:
        print(
            f"Warning: Only {len(filtered_questions)} question(s) available for age group '{age_group}'.")
        return filtered_questions

    # Return random selection
    return random.sample(filtered_questions, count)


def get_available_age_groups(data):
    """Get all unique age groups in the dataset."""
    if not data:
        return []
    return list(set(item.get("age_group", "Unknown") for item in data))


def display_questions(questions, language="ar"):
    """
    Display questions in a formatted way.

    Parameters:
    -----------
    questions : list
        List of question dictionaries
    language : str
        "en" for English, "ar" for Arabic, "both" for bilingual
    """

    if not questions:
        print("No questions to display.")
        return

    print("\n" + "=" * 60)
    print(f"🎯 RANDOM QUESTIONS ({len(questions)} selected)")
    print("=" * 60)

    for i, q in enumerate(questions, 1):
        print(f"\n📌 Question {i}:")
        print(f"   Age Group: {q.get('age_group', 'N/A')}")

        if language in ["en", "both"]:
            print(f"\n   🇬🇧 English:")
            print(f"   {q.get('question_en', 'N/A')}")
            print("   Options:")
            for j, opt in enumerate(q.get('options_en', []), 1):
                print(f"      {j}. {opt}")

        if language in ["ar", "both"]:
            print(f"\n   🇸🇦 Arabic:")
            print(f"   {q.get('question_ar', 'N/A')}")
            print("   الخيارات:")
            for j, opt in enumerate(q.get('options_ar', []), 1):
                print(f"      {j}. {opt}")

        print(f"\n   🧠 Concepts: {', '.join(q.get('concepts_en', []))}")
        print("-" * 60)
```

```python


questions = load_questions("questions_en_ar.json")

if questions:
    # Example 1: Get 2 random questions for age 6-9
    print("\n🔹 Example 1: Age 6-9")
    random_questions = get_random_questions_by_age(
        questions, "6-9", count=2)
    display_questions(random_questions, language="ar")

    # Example 2: Get 2 random questions for age 10-12
    print("\n🔹 Example 2: Age 10-14")
    random_questions = get_random_questions_by_age(
        questions, "10-14", count=2)
    display_questions(random_questions, language="ar")

    # Example 3: Get 3 random questions for age 15-18
    print("\n🔹 Example 3: Age 15-18    ")
    random_questions = get_random_questions_by_age(
        questions, "15-18", count=2)
    display_questions(random_questions, language="ar")

    # Example 4: Show available age groups
    print("\n📂 Available Age Groups:")
    print(get_available_age_groups(questions))
```
